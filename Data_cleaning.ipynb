```python
# converting total charges to numeric and changing blank entries to NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
#  Converting SeniorCitizen values to Yes/No labels and forming a new column
df['SeniorCitizen_Label'] = df['SeniorCitizen'].map({0: 'No', 1: 'Yes'})
# Display the original and labeled columns
df[['SeniorCitizen', 'SeniorCitizen_Label']].head()
```

```python
# Converting Churn to 0/1 values
df['Churn_Flag'] = df['Churn'].map({'No': 0, 'Yes': 1})
```

```python
# Count each churn category
churn_counts = df['Churn'].value_counts()

# Calculate the churn rate
churn_rate = df['Churn_Flag'].mean() * 100

# Plot the churn distribution
plt.figure(figsize=(5, 5))
plt.pie(churn_counts, labels=churn_counts.index, autopct='%1.1f%%',
        colors=['#4CAF50', '#F44336'], startangle=90)
plt.title("Overall Customer Churn Distribution")
plt.show()

# Print the overall churn rate
print(f"Overall churn rate: {churn_rate:.2f}%")
```

```python
# Define a function to plot churn rate by category
def churn_rate_by(col, rotate=0):
    rate = df.groupby(col)['Churn_Flag'].mean().sort_values(ascending=False) * 100
    ax = rate.plot(kind='bar', color='#4C72B0', edgecolor='black')
    plt.title(f"Churn Rate by {col}")
    plt.ylabel("Churn Rate (%)")
    plt.xlabel(col)
    plt.xticks(rotation=rotate)
    for i, v in enumerate(rate):
        ax.text(i, v + 0.5, f"{v:.1f}%", ha='center')
    plt.tight_layout()
    plt.show()
    return rate

# Plot churn rate by contract type
churn_rate_by('Contract')
```

```python
# Churn rate by internet servide
churn_rate_by("InternetService")
```

```python
#Churn rate by payment method
churn_rate_by("PaymentMethod",rotate=20)
```

```python
# Churn rate across key add-on services (Online Security, Tech Support, etc.)
addon_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, col in zip(axes.flatten(), addon_cols):
    rate = df.groupby(col)['Churn_Flag'].mean().sort_values(ascending=False) * 100
    rate.plot(kind='bar', ax=ax, color='#55A868', edgecolor='black')
    ax.set_title(f"Churn Rate by {col}")
    ax.set_ylabel("Churn Rate (%)")
    ax.set_xlabel("")
    ax.tick_params(axis='x', rotation=20)
plt.tight_layout()
plt.show()
```

```python
#---------BUSINESS QUESTIONS-----
```

```python
# Q1 What is the overall churn rate?
overall_churn_rate = df['Churn_Flag'].mean() * 100
print(f"Overall churn rate: {overall_churn_rate:.2f}%")
```

```python
'''Answer:About 26.54% of customers in this dataset have churned — roughly 1 in 4 customers,
 which is a significant retention concern for the business.'''
```

```python
#Q2. Which contract type has the highest churn, and what does that suggest about business strategy?
contract_churn = df.groupby('Contract')['Churn_Flag'].mean().sort_values(ascending=False) * 100
print(contract_churn)

# Answer: Month-to-month contracts have by far the highest churn rate, while 1-year and
# especially 2-year contracts show much stronger retention. This suggests the business
# should incentivize customers to move from month-to-month to longer-term contracts
# (e.g. via discounts or bundled perks) to improve retention.
```

```python
# Q3. Does customer tenure influence the likelihood of churn?
tenure_by_churn = df.groupby('Churn')['tenure'].mean()
print(tenure_by_churn)

# Answer: Yes — churned customers have a much lower average tenure than retained customers.
# Churn risk is heavily front-loaded into the first several months of the customer
# relationship, making onboarding and early engagement critical.
```

```python
# Q4. Which payment method is associated with the highest churn?
payment_churn = df.groupby('PaymentMethod')['Churn_Flag'].mean().sort_values(ascending=False) * 100
print(payment_churn)

# Answer: 'Electronic check' users churn at the highest rate, well above customers using
# automatic bank transfer or credit card payments. Encouraging a shift to automatic
# payment methods could reduce friction-driven churn.
```

```python
# Q5. Does Internet Service type affect churn?
internet_churn = df.groupby('InternetService')['Churn_Flag'].mean().sort_values(ascending=False) * 100
print(internet_churn)

# Answer: Fiber optic customers churn considerably more than DSL customers or those with no internet service.
```

```python
# Q6. Do add-on services (Online Security, Tech Support, etc.) help reduce churn?
for col in addon_cols:
    rate = df.groupby(col)['Churn_Flag'].mean().sort_values(ascending=False) * 100
    print(f"\n{col}:")
    print(rate)

# Answer: Yes — customers without these add-on services (answer = 'No') consistently show
# higher churn than those who have them. TechSupport and OnlineSecurity show the largest
```

```python
# Q7. Are Senior Citizens more likely to churn?
senior_churn = df.groupby('SeniorCitizen_Label')['Churn_Flag'].mean() * 100
print(senior_churn)

# Answer: Yes — senior citizens churn at a noticeably higher rate than non-seniors,
# indicating a need for senior-focused retention or support programs.
```

```python
# Q8. Does having a Partner or Dependents reduce churn?
partner_churn = df.groupby('Partner')['Churn_Flag'].mean() * 100
dependents_churn = df.groupby('Dependents')['Churn_Flag'].mean() * 100
print("By Partner status:\n", partner_churn)
print("\nBy Dependents status:\n", dependents_churn)

# Answer: Yes on both counts,customers without a partner or without dependents churn
# more. Household/family ties appear to correlate with more stable, long-term subscriptions.
```

```python
# Q9. How much monthly revenue is at risk from churned customers?
monthly_revenue_at_risk = df.loc[df['Churn'] == 'Yes', 'MonthlyCharges'].sum()
total_monthly_revenue = df['MonthlyCharges'].sum()
pct_at_risk = monthly_revenue_at_risk / total_monthly_revenue * 100

print(f"Monthly revenue from churned customers: ${monthly_revenue_at_risk:,.2f}")
print(f"Total monthly revenue (all customers): ${total_monthly_revenue:,.2f}")
print(f"Percentage of monthly revenue at risk: {pct_at_risk:.2f}%")

# Answer: Churned customers represent a substantial chunk of monthly recurring revenue
# lost, underscoring the direct financial impact of the ~26.5% churn rate.
```

```python
# Q10. Which customer segment is the highest priority for retention efforts?
high_risk_segment = df[
    (df['Contract'] == 'Month-to-month') &
    (df['InternetService'] == 'Fiber optic') &
    (df['PaymentMethod'] == 'Electronic check')
]

high_risk_churn_rate = high_risk_segment['Churn_Flag'].mean() * 100
segment_size = len(high_risk_segment)

print(f"High-risk segment size: {segment_size} customers")
print(f"Churn rate within this segment: {high_risk_churn_rate:.2f}%")
print(f"Compare to overall churn rate: {overall_churn_rate:.2f}%")

# Answer: Customers who combine month-to-month contracts, fiber optic internet, and
# electronic check payments churn at a much higher rate than the overall base. This
# segment should be the top priority for targeted retention campaigns (e.g. contract
# upgrade offers, payment method incentives, proactive customer support outreach).
```
